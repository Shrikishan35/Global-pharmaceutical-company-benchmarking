**Table identification note:** Product-revenue tables are selected using product names, revenue labels, U.S./Outside U.S. headings and fiscal-year labels. The notebook no longer depends on fixed HTML table positions.

```python
# Identify product-revenue tables by their contents, not their position
import re

html_tables = soup.find_all("table")


def find_product_revenue_table(tables, required_products, years):
    """Return the table containing the requested products and geographic revenue headings."""
    matches = []

    for table_index, table in enumerate(tables):
        table_text = re.sub(r"\s+", " ", table.get_text(" ", strip=True)).lower()
        normalized_text = table_text.replace("outside the u.s.", "outside u.s.")

        has_products = all(product.lower() in normalized_text for product in required_products)
        has_revenue_label = "revenue" in normalized_text
        has_us_geography = "u.s." in normalized_text and "outside u.s." in normalized_text
        has_years = all(str(year) in normalized_text for year in years)

        if has_products and has_revenue_label and has_us_geography and has_years:
            matches.append((table_index, table))

    if len(matches) != 1:
        raise ValueError(
            f"Expected exactly one product-revenue table; found {len(matches)}. "
            "Review the table headings and required product names."
        )

    return matches[0]


product_table_index, product_table = find_product_revenue_table(
    html_tables,
    required_products=["Mounjaro", "Trulicity", "Verzenio"],
    years=[2025, 2024, 2023]
)

print("Total HTML tables:", len(html_tables))
print("Product-revenue table identified by content. Table index:", product_table_index)
if product_table_index != 29:
    raise AssertionError(f"Expected FY2025 product table index 29; found {product_table_index}. Review the identified table before continuing.")
print(product_table.get_text(" ", strip=True)[:1200])
```

```python
# Extract selected product revenue from the SEC filing

selected_products = [
    "Mounjaro",
    "Zepbound (1)",
    "Trulicity",
    "Jardiance (2)",
    "Verzenio",
    "Taltz"
]

extracted_products = []

for row in product_table.find_all("tr"):
    cells = [
        cell.get_text(" ", strip=True)
        for cell in row.find_all(["th", "td"])
    ]

    cells = [value for value in cells if value]

    if cells and cells[0] in selected_products:
        values = []

        for value in cells[1:]:
            if value == "$":
                continue
            elif value == "—":
                values.append(0)
            else:
                values.append(float(value.replace(",", "")))

        if len(values) == 6:
            extracted_products.append([cells[0]] + values)

product_df = pd.DataFrame(
    extracted_products,
    columns=[
        "Product",
        "US_2025", "US_2024", "US_2023",
        "Outside_US_2025", "Outside_US_2024", "Outside_US_2023"
    ]
)

print(product_df.to_string(index=False))
```

```python
# Calculate worldwide product revenue (USD millions)

for year in [2023, 2024, 2025]:
    product_df[f"Worldwide_{year}"] = (
        product_df[f"US_{year}"]
        + product_df[f"Outside_US_{year}"]
    )

worldwide_revenue = product_df[
    ["Product", "Worldwide_2023", "Worldwide_2024", "Worldwide_2025"]
]

print(worldwide_revenue.to_string(index=False))
```

```python
# Calculate year-over-year product revenue growth

for year in [2024, 2025]:
    previous_year = year - 1

    worldwide_revenue[f"Growth_{year}_%"] = (
        (worldwide_revenue[f"Worldwide_{year}"]
         - worldwide_revenue[f"Worldwide_{previous_year}"])
        / worldwide_revenue[f"Worldwide_{previous_year}"] * 100
    )

print(worldwide_revenue.round(2).to_string(index=False))
```

```python
# Product contribution within the six-product comparison set only
# This is NOT contribution to total company revenue growth.

worldwide_revenue["Revenue_Increase_2025_USD_mn"] = (
    worldwide_revenue["Worldwide_2025"]
    - worldwide_revenue["Worldwide_2024"]
)

selected_products_net_increase = worldwide_revenue[
    "Revenue_Increase_2025_USD_mn"
].sum()

worldwide_revenue["Contribution_to_Selected_Product_Net_Growth_%"] = (
    worldwide_revenue["Revenue_Increase_2025_USD_mn"]
    / selected_products_net_increase * 100
)

print(
    worldwide_revenue[
        ["Product", "Revenue_Increase_2025_USD_mn",
         "Contribution_to_Selected_Product_Net_Growth_%"]
    ].round(2).to_string(index=False)
)
```

### Eli Lilly — Financial Analysis: Methodology and Limitations

**Analysis period:** FY2021–FY2025. FY2020 revenue is retained for historical context only; the primary CAGR uses FY2021–FY2025 (four intervals).

**Data sources:** Eli Lilly's SEC Company Facts data and annual Form 10-K filings. Cash-flow references: FY2023 Form 10-K, Consolidated Statements of Cash Flows (FY2021–FY2023), https://www.sec.gov/Archives/edgar/data/59478/000005947824000065/lly-20231231.htm; FY2025 Form 10-K, Consolidated Statements of Cash Flows (FY2024–FY2025), https://www.sec.gov/Archives/edgar/data/59478/000005947826000013/lly-20251231.htm. Financial and product-level figures were extracted from company disclosures. R&D expenditure figures were transcribed from the consolidated statements of operations in Lilly annual Form 10-K filings because the standard SEC XBRL concept did not provide a complete five-year series. The FY2024 Form 10-K independently confirms FY2022–FY2024 R&D values in Item 8, Consolidated Statements of Operations. The manually entered COGS, SG&A, net income, and R&D figures are transcribed from the Consolidated Statements of Operations in the applicable annual Form 10-K; the FY2024 filing reports FY2022–FY2024 comparative values.

**Cash-flow source clarification:** The FY2021 and FY2022 values previously copied from the SEC Company Facts extraction (7.2607 and 7.0844 USD bn) did not match the audited consolidated cash flow statements. They have been replaced with the Form 10-K values (7.3659 and 7.5857 USD bn). Capital expenditure is the reported line item “Purchases of property and equipment”; FY2024 is 5.0578 USD bn.

**Calculation methodology:**
- Year-over-year revenue growth = (Current-year revenue / Previous-year revenue − 1) × 100.
- CAGR = (Ending value / Beginning value)^(1 / Number of intervals) − 1.
- Gross margin = (Revenue − Cost of sales) / Revenue × 100.
- Net profit margin = Net income / Revenue × 100.
- R&D intensity = R&D expenditure / Revenue × 100.
- Free cash flow = Operating cash flow − Capital expenditure.

**Comparability and limitations:**
1. Operating margin has not been included pending a consistent definition and verification of operating income across all six companies. Do not substitute net income or operating cash flow. Net profit margin and operating cash flow margin are reported separately and are not substitutes for operating margin.
2. FY2021–FY2025 represents five financial years and four CAGR intervals.
3. Product and therapeutic-area disclosures differ between annual filings. FY2023 figures were cross-checked between the FY2023 and FY2025 filings; minor differences reflect rounding.
4. Neuroscience and other product categories have changes in disclosure aggregation. Their long-term growth should therefore be interpreted cautiously.
5. Product-level analysis focuses on six selected products with comparable disclosures across filings; it does not represent the entire product portfolio.
6. FY2025 growth was strongly influenced by Mounjaro and Zepbound. Their combined revenue and contribution to growth should be interpreted alongside the wider portfolio and company disclosures.

**Interpretation:** Eli Lilly's recent growth is increasingly concentrated in its cardiometabolic portfolio, particularly tirzepatide-based products. The financial analysis also shows improving gross margin and higher capital expenditure. These findings will be compared with Pfizer and the wider pharmaceutical segment before drawing competitive conclusions.

**Neuroscience comparability check (FY2024 Form 10-K):** The FY2024 Form 10-K reports total neuroscience revenue of approximately USD 1,473.4 million in FY2024 and USD 2,878.5 million in FY2023. This is consistent with the FY2025 filing's comparative FY2023 disclosure (rounding differences aside). The FY2023 neuroscience figure includes approximately USD 1,694.8 million from Zyprexa, including a sale of rights for the olanzapine portfolio in July 2023, as disclosed in the FY2024 Form 10-K footnote. Therefore, the FY2023–FY2024 decline is materially affected by this non-recurring rights transaction and should not be interpreted as an equivalent decline in underlying neuroscience demand. Other neuroscience revenue was approximately USD 505.5 million in FY2023 and USD 486.7 million in FY2024. Keep the neuroscience and other category growth rates out of headline presentation claims unless this disclosure effect is explicitly explained. ```python

```python
# Build full disclosed product revenue dataset (FY2021–FY2025)

import pandas as pd
import re

# 1. Clean product names and remove filing footnote markers
historical_all = historical_clean.copy()
recent_all = recent_clean.copy()

for df in [historical_all, recent_all]:
    df["Product_Clean"] = (
        df["Product"]
        .str.replace(r"\s*\(\d+\)", "", regex=True)
        .str.replace("®", "", regex=False)
        .str.strip()
    )

# 2. Consolidate duplicate product names within each filing
historical_all = (
    historical_all
    .groupby("Product_Clean", as_index=False)
    .agg({
        "Worldwide_2021": "sum",
        "Worldwide_2022": "sum"
    })
)

# Use the latest filing's FY2023 comparative values
# and FY2024–FY2025 values
for year in [2023, 2024, 2025]:
    recent_all[f"Worldwide_{year}"] = (
        recent_all[f"US_{year}"]
        + recent_all[f"Outside_US_{year}"]
    )

recent_all = (
    recent_all
    .groupby("Product_Clean", as_index=False)
    .agg({
        "Worldwide_2023": "sum",
        "Worldwide_2024": "sum",
        "Worldwide_2025": "sum"
    })
)

# 3. Combine the historical and latest disclosures
lilly_all_products = historical_all.merge(
    recent_all,
    on="Product_Clean",
    how="outer"
)

# 4. Arrange columns chronologically
lilly_all_products = lilly_all_products[
    [
        "Product_Clean",
        "Worldwide_2021",
        "Worldwide_2022",
        "Worldwide_2023",
        "Worldwide_2024",
        "Worldwide_2025"
    ]
]

# 5. Sort by FY2025 revenue
lilly_all_products = lilly_all_products.sort_values(
    "Worldwide_2025",
    ascending=False,
    na_position="last"
).reset_index(drop=True)

# 6. Inspect results
print("Total disclosed product/category rows:",
      len(lilly_all_products))

display(lilly_all_products.round(2))

# 7. Export
output_path = Path(
    "../data/processed/eli_lilly_full_product_revenue_2021_2025.csv"
)

lilly_all_products.to_csv(output_path, index=False)

print("Saved:", output_path)
```
